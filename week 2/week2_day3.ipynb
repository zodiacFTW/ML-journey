```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X,Y=load_iris(return_X_y=True)
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

dt_gini=DecisionTreeClassifier(criterion='gini',max_depth=3,random_state=42)
dt_entropy=DecisionTreeClassifier(criterion='entropy',max_depth=3,random_state=42)

dt_gini.fit(X_train,Y_train)
dt_entropy.fit(X_train,Y_train)

plt.figure(figsize=(16,6))

plot_tree(dt_gini,feature_names=load_iris().feature_names,class_names=load_iris().target_names,filled=True,rounded=True,fontsize=9)
plt.title("decision tree (Gini)")
plt.show()
```

```python
def gini(y):
    #L=summation(p(1-p))=1-summation p^2
    classes,counts=np.unique(y,return_counts=True)
    p=counts/len(y)
    return 1-np.sum(p**2)

def entropy(y):
    #l=-sigma(plogp)
    classes,counts=np.unique(y,return_counts=True)
    p=counts/len(y)
    return -np.sum(p*np.log2(p+1e-9))

```
